```python
# make a dictionary containing the number of times each character appears

text = "mississippi"

count = {}

letters = list(text)
print(letters)

for letter in letters:
    if letter in count:
        count[letter] += 1
    else:
        count[letter] = 1

print(count)


# even better solution

for letter in letters:
    count[letter] = count.get(letter, 0) + 1

print(count)
```

```python
# word frequency 

import string

sentence = "The cat sat on the mat and the cat ate the rat."

no_punctuation = ''.join(char for char in sentence if char not in string.punctuation)

print(no_punctuation)

frequency_dic = {}

words = no_punctuation.split()

lowercase_words = [word.lower() for word in words]

for word in lowercase_words:
    if word in frequency_dic:
        frequency_dic[word] += 1
    else:
        frequency_dic[word] = 1

print(frequency_dic)       
```

```python
# group numbers by remainder 

numbers = [12, 7, 5, 19, 22, 14, 8, 31, 17, 4]

newdic = {}

for number in numbers:
    remainder = number % 3
    if remainder not in newdic:
        newdic[remainder] = []
    newdic[remainder].append(number)

print(newdic)
```

```python
# first unique character. find the first character that appears exactly once

text = "aabbccddeeffgxxyy"

counter = {}

letters = list(text)

for letter in letters:
    if letter not in counter:
        counter[letter] = 1
    else:
        counter[letter] += 1

for letter, count in counter.items():
    if count == 1:
        unique_character = letter
        break
        

print(unique_character)
```

```python
# Coke Machine

amount_due = 50

accepted_coins = [25, 10, 5]

attempts = 0

while attempts < 5:
    try:
        print(f"Amount Due: {amount_due}")
        coin_input = int(input("Insert Coin:"))
        if coin_input in accepted_coins:
            amount_due = amount_due - coin_input
            attempts += 1
            if amount_due < 0:
                change = abs(amount_due)
        else:
            pass
    except ValueError:
        pass

print(f"Change Due: {change}")
```

```python
# Vanity plates. 

import string

def main():
    plate = str(input("Plate:"))
    if isValid(plate):
        print("Valid")
    else:
        print("Invalid")



def isValid(input):
    if len(input) < 2 and len(input) > 6:
        return False
    elif input[0].isalpha() == False and input[1].isalpha() == False:
        return False
    number_started = False
    for char in input:
        if char.isnumeric():
            number_started = True
        elif number_started:
            return False
    else:
        return True

if __name__ == "__main__":
    main()
```
